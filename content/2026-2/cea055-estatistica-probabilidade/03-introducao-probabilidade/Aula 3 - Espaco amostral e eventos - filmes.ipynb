# 🎲 Estatística na prática: espaço amostral e eventos

Duas partes neste notebook: primeiro simulamos o exemplo clássico dos dados que vimos na aula, depois aplicamos a mesma ideia de espaço amostral e evento no catálogo de filmes. Só executar e ler, sem entrega.

## Parte 1 — o exemplo dos dois dados, simulado

Na aula vimos: lançam-se dois dados de 6 faces, e definimos eventos como "faces iguais" ou "soma igual a 10". Vamos simular isso de verdade em Python.

```python
import random

random.seed(42)  # fixa a semente só para o resultado ser sempre o mesmo neste notebook

# espaço amostral: todos os pares possíveis (dado1, dado2), cada um de 1 a 6
espaco_amostral = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
print(f"O espaço amostral tem {len(espaco_amostral)} elementos:")
print(espaco_amostral)
```

Agora vamos definir os mesmos três eventos vistos na aula, como **subconjuntos** desse espaço amostral:

```python
evento_A = [(d1, d2) for (d1, d2) in espaco_amostral if d1 == d2]
evento_B = [(d1, d2) for (d1, d2) in espaco_amostral if d1 + d2 == 10]
evento_C = [(d1, d2) for (d1, d2) in espaco_amostral if d1 + d2 < 2]

print(f"A (faces iguais): {evento_A}")
print(f"P(A) = {len(evento_A)}/{len(espaco_amostral)} = {len(evento_A)/len(espaco_amostral):.3f}")
print()
print(f"B (soma igual a 10): {evento_B}")
print(f"P(B) = {len(evento_B)}/{len(espaco_amostral)} = {len(evento_B)/len(espaco_amostral):.3f}")
print()
print(f"C (soma menor que 2): {evento_C}")
print(f"P(C) = {len(evento_C)}/{len(espaco_amostral)} = {len(evento_C)/len(espaco_amostral):.3f}")
```

**Repare** que o evento C é o conjunto vazio — não existe jeito de somar dois dados (cada um de 1 a 6) e dar menos que 2. Faz sentido: P(C) = 0.

### Conferindo por simulação

Em vez de contar os elementos do espaço amostral, vamos lançar os dois dados **de verdade** (com o computador) 100.000 vezes e ver se a frequência relativa do evento A bate com a probabilidade teórica que calculamos acima.

```python
lancamentos = 100_000
contagem_A = 0

for _ in range(lancamentos):
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    if d1 == d2:
        contagem_A += 1

frequencia_relativa_A = contagem_A / lancamentos
print(f"Frequência relativa do evento A em {lancamentos} lançamentos simulados: {frequencia_relativa_A:.3f}")
print(f"Probabilidade teórica calculada acima: {len(evento_A)/len(espaco_amostral):.3f}")
```

Os dois valores ficam bem próximos — essa é a ideia por trás da definição de probabilidade como frequência relativa "no longo prazo", que a definição de dicionário do início da aula menciona.

## Parte 2 — a mesma ideia aplicada ao catálogo de filmes

Agora vamos tratar "escolher um filme aleatoriamente do catálogo" como um experimento aleatório, e ver como espaço amostral e evento aparecem num dado real.

```python
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    "titulo": ["Vingadores: Ultimato", "Toy Story", "O Poderoso Chefão", "Coringa",
               "Cidade de Deus", "Interestelar", "A Origem", "Shrek",
               "O Iluminado", "Parasita", "Bacurau", "Divertida Mente",
               "Duna", "Homem-Aranha: Sem Volta Para Casa", "Get Out - Corra!", "Klaus",
               "Tropa de Elite", "Zumbilândia", "Whiplash", "Coco",
               "Corra Que a Policia Vem Ai", "Não Olhe Para Cima", "Barbie", "Oppenheimer"],
    "genero": ["Ação", "Animação", "Drama", "Drama",
               "Drama", "Ficção Científica", "Ficção Científica", "Animação",
               "Terror", "Drama", "Ficção Científica", "Animação",
               "Ficção Científica", "Ação", "Terror", "Animação",
               "Ação", "Comédia", "Drama", "Animação",
               "Comédia", "Comédia", "Comédia", "Drama"],
    "classificacao_indicativa": ["12", "Livre", "16", "16",
                                  "18", "10", "12", "Livre",
                                  "16", "16", "16", "Livre",
                                  "12", "12", "16", "Livre",
                                  "18", "14", "12", "Livre",
                                  "12", "12", "10", "14"],
    "ano": [2019, 1995, 1972, 2019, 2002, 2014, 2010, 2001, 1980, 2019,
            2019, 2015, 2021, 2021, 2017, 2019, 2007, 2009, 2014, 2017,
            1981, 2021, 2023, 2023],
    "duracao_minutos": [181, 81, 175, 122, 130, 169, 148, 90, 146, 132,
                         131, 95, 155, 148, 104, 96, 115, 88, 107, 105,
                         94, 138, 114, 180],
    "nota_publico": [8.4, 8.3, 9.2, 8.4, 8.6, 8.7, 8.8, 7.9, 8.4, 8.6,
                      7.0, 8.1, 8.0, 8.3, 7.7, 7.5, 7.7, 7.6, 8.5, 8.4,
                      7.7, 7.2, 6.9, 8.4],
    "bilheteria_milhoes": [2797, 373, 246, 1074, 30, 701, 836, 484, 47, 258,
                            1.2, 858, 434, 1922, 255, 3.4, 25, 75, 13, 814,
                            36, 787, 1447, 950],
}

df = pd.DataFrame(dados)
df.head(10)
```

Aqui, o **espaço amostral** é o conjunto dos 24 filmes do catálogo. Um **evento** é qualquer subconjunto — por exemplo, "o filme sorteado é do gênero Drama".

```python
evento_drama = df[df["genero"] == "Drama"]

print(f"Espaço amostral: {len(df)} filmes")
print(f"Evento 'é Drama': {len(evento_drama)} filmes")
print(f"P(Drama) = {len(evento_drama)}/{len(df)} = {len(evento_drama)/len(df):.3f}")
evento_drama[["titulo", "genero"]]
```

### Combinando dois eventos (interseção)

```python
evento_drama_e_livre = df[(df["genero"] == "Drama") & (df["classificacao_indicativa"] == "Livre")]

print(f"P(Drama E Livre) = {len(evento_drama_e_livre)}/{len(df)} = {len(evento_drama_e_livre)/len(df):.3f}")
evento_drama_e_livre[["titulo", "genero", "classificacao_indicativa"]]
```

O resultado dá zero — não existe, no nosso catálogo, nenhum filme de Drama classificado como Livre. Isso é só uma característica dos nossos 24 filmes de exemplo, não uma regra geral.

### Conferindo por simulação, de novo

```python
sorteios = df.sample(n=100_000, replace=True, random_state=42)
frequencia_relativa_drama = (sorteios["genero"] == "Drama").mean()

print(f"Frequência relativa de 'Drama' em 100.000 sorteios simulados: {frequencia_relativa_drama:.3f}")
print(f"Probabilidade teórica: {len(evento_drama)/len(df):.3f}")
```

## Para fechar

Repare que os dois pedaços deste notebook usam exatamente a mesma lógica: contar quantos elementos do espaço amostral satisfazem uma condição, e dividir pelo total. Isso vale tanto para dois dados quanto para um catálogo de filmes, um baralho de cartas, ou uma base de dados com milhões de linhas — a definição de probabilidade não muda, só o tamanho do espaço amostral.
