## 1. As notas do público são parecidas ou espalhadas?

### A situação e a pergunta
Um serviço de streaming quer saber se o catálogo é **homogêneo**: os filmes têm notas parecidas ou há muita diferença entre eles? Saber só a média não responde isso.

### A ferramenta da aula
As medidas de **dispersão**, que medem o quanto os dados se afastam da média $\bar{x}$:

- **Desvio médio:** a média das distâncias até a média, $\;DM = \frac{1}{n}\sum |x_i - \bar{x}|$;
- **Variância:** a média dos **quadrados** dessas distâncias, $\;\sigma^2 = \frac{1}{n}\sum (x_i - \bar{x})^2$;
- **Desvio padrão:** a raiz da variância, $\;\sigma = \sqrt{\sigma^2}$, que volta para a **mesma unidade** dos dados (pontos de nota).

### O que o código faz
Segue as fórmulas passo a passo:
1. calcula a média das notas;
2. calcula o **desvio** de cada filme (nota − média);
3. tira a média dos desvios em valor absoluto (desvio médio) e dos desvios ao quadrado (variância);
4. tira a raiz da variância (desvio padrão).

```python
nota = df["nota_publico"]                  # a coluna de notas

media = nota.mean()                        # passo 1: média
desvios = nota - media                     # passo 2: desvio de cada filme
desvio_medio = desvios.abs().mean()        # passo 3: média das distâncias (abs = valor absoluto)
variancia = (desvios ** 2).mean()          #          média dos desvios ao quadrado
desvio_padrao = variancia ** 0.5           # passo 4: raiz quadrada da variância

print(f"Média:         {media:.2f}")
print(f"Desvio médio:  {desvio_medio:.2f}")
print(f"Variância:     {variancia:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")
```

**Como ler:** a nota média é **8,10**, e os filmes se afastam dela, tipicamente, cerca de **0,5 ponto** (desvio médio 0,48; desvio padrão 0,57). Numa escala de 0 a 10, é pouco: o catálogo é bem **homogêneo**, quase todos os filmes são bem avaliados. A variância (0,33) está em "pontos ao quadrado", por isso é mais difícil de interpretar diretamente; o desvio padrão resolve isso.

> **Detalhe técnico:** o `pandas` tem funções prontas (`.var()` e `.std()`), mas elas dividem por $n-1$ em vez de $n$ (é a variância **amostral**, que aparece mais adiante no curso). Por isso fizemos a conta "à mão", para bater com a definição da aula.

## 2. Quartis: onde fica a "metade do meio" do catálogo?

### A pergunta
A partir de que nota um filme está entre os **25% mais bem avaliados**? E entre quais notas fica a **metade central** do catálogo?

### A ferramenta da aula
Os **quartis** dividem os dados **ordenados** em quatro partes com a mesma quantidade de filmes:
- **Q1** (1º quartil): 25% dos filmes têm nota até ele;
- **Q2** (mediana): 50%;
- **Q3** (3º quartil): 75%.

A **distância interquartil** $\;DIQ = Q_3 - Q_1\;$ é a largura da metade central dos dados, uma medida de dispersão que **não se deixa levar por valores extremos**.

### O que o código faz
Pede ao `pandas` os quantis 0,25, 0,50 e 0,75 da nota (`quantile`) e calcula a DIQ.

```python
q1 = nota.quantile(0.25)     # 25% das notas estão abaixo
q2 = nota.quantile(0.50)     # mediana
q3 = nota.quantile(0.75)     # 75% das notas estão abaixo
diq = q3 - q1                # largura da metade central

print(f"Q1:        {q1:.2f}")
print(f"Mediana:   {q2:.2f}")
print(f"Q3:        {q3:.2f}")
print(f"DIQ:       {diq:.2f}")
```

**Como ler:** a metade central do catálogo tem notas entre **7,70 e 8,43** (DIQ ≈ 0,73). Um filme com nota **acima de 8,43** está entre os 25% mais bem avaliados. A mediana (8,30) é um pouco maior que a média (8,10): alguns filmes com nota baixa puxam a média para baixo.

> **Detalhe técnico:** existem vários jeitos de calcular quartis, e com poucos dados eles podem dar valores ligeiramente diferentes. O `pandas` usa **interpolação** (por isso aparece 8,425, que não é nota de nenhum filme). Se a sua conta à mão pelo método da aula der algo um pouco diferente, não é erro.

## 3. Boxplot: existe algum filme "fora da curva"?

### A pergunta
Há algum filme com nota **atípica** (um **outlier**), muito acima ou muito abaixo do resto?

### A ferramenta da aula
O **boxplot** desenha os quartis:
- a **caixa** vai de Q1 a Q3, e a linha no meio é a mediana;
- os **bigodes** vão até o dado mais extremo que ainda está dentro dos limites $\;Q_1 - 1{,}5 \cdot DIQ\;$ e $\;Q_3 + 1{,}5 \cdot DIQ$;
- quem fica **fora** desses limites aparece como um ponto isolado: é um **outlier**.

### A resposta pela conta
Limites: $\;7{,}70 - 1{,}5 \times 0{,}725 \approx 6{,}61\;$ e $\;8{,}425 + 1{,}5 \times 0{,}725 \approx 9{,}51$. A menor nota é 6,9 e a maior é 9,2: ambas **dentro** dos limites.

### O que o código faz
Desenha dois boxplots lado a lado: o da **nota** e o da **bilheteria**, para comparar.

```python
fig, eixos = plt.subplots(1, 2, figsize=(10, 4))      # uma figura com 2 gráficos lado a lado

eixos[0].boxplot(nota)
eixos[0].set_title("Nota do público")

eixos[1].boxplot(df["bilheteria_milhoes"])
eixos[1].set_title("Bilheteria (US$ milhões)")

plt.tight_layout()                                     # ajusta o espaçamento
plt.show()
```

**Como ler:**
- **Nota (à esquerda):** nenhum ponto isolado, então **não há outlier**, como a conta mostrou. A caixa é curta: as notas estão concentradas.
- **Bilheteria (à direita):** a caixa fica "espremida" embaixo e há **um ponto isolado lá em cima**: é *Vingadores: Ultimato* (cerca de 2,8 bilhões de dólares), um outlier. O bigode de cima é bem mais comprido que o de baixo, um sinal de **assimetria**, que é o tema da próxima seção.

## 4. Simetria: média ou mediana para resumir a bilheteria?

### A pergunta
Para dizer quanto um filme do catálogo fatura "tipicamente", devemos usar a **média** ou a **mediana** da bilheteria?

### A ferramenta da aula
- Numa distribuição **simétrica**, média e mediana ficam **próximas**;
- numa distribuição **assimétrica à direita** (cauda longa para valores altos), poucos valores enormes **puxam a média para cima**, e a mediana representa melhor o valor típico.

### O que o código faz
1. desenha os **histogramas** da nota e da bilheteria, lado a lado, para ver o formato de cada distribuição;
2. calcula a média e a mediana das duas variáveis.

```python
fig, eixos = plt.subplots(1, 2, figsize=(11, 4))

eixos[0].hist(nota, bins=6, color="steelblue", edgecolor="black")               # 6 faixas de nota
eixos[0].set_title("Nota do público")
eixos[0].set_xlabel("Nota")

eixos[1].hist(df["bilheteria_milhoes"], bins=6, color="coral", edgecolor="black")
eixos[1].set_title("Bilheteria")
eixos[1].set_xlabel("Bilheteria (US$ milhões)")

plt.tight_layout()
plt.show()

bilheteria = df["bilheteria_milhoes"]
print(f"Nota       — média: {nota.mean():7.2f}   mediana: {nota.median():7.2f}")
print(f"Bilheteria — média: {bilheteria.mean():7.1f}   mediana: {bilheteria.median():7.1f}   (US$ milhões)")
```

**Como ler:**
- **Nota:** as barras se espalham sem uma cauda longa, e média (8,10) e mediana (8,30) ficam **próximas**.
- **Bilheteria:** a maioria dos filmes está nas primeiras barras, e poucos blockbusters formam uma **cauda longa à direita**. A média (**602,8 milhões**) é **50% maior** que a mediana (**403,5 milhões**).

**Resposta:** para a bilheteria, a **mediana** representa melhor o filme "típico": metade dos filmes fatura menos de 403,5 milhões. O mesmo vale para renda, preço de imóveis e salários, que também são assimétricos à direita.

## 5. Qual gênero tem as notas mais altas e mais consistentes?

### A pergunta
Algum gênero se destaca? E em qual deles as notas **variam mais**?

### A ferramenta da aula
**Boxplots lado a lado**, um por grupo: a **altura da mediana** compara o nível das notas, e o **tamanho da caixa e dos bigodes** compara a dispersão.

### O que o código faz
Desenha um boxplot da nota para **cada gênero** (`by="genero"` = "separado por gênero").

```python
df.boxplot(column="nota_publico", by="genero", figsize=(9, 5), rot=30)   # rot=30 inclina os nomes
plt.title("Nota do público por gênero")
plt.suptitle("")                       # remove um título automático que o pandas coloca
plt.ylabel("Nota do público")
plt.show()

print(df.groupby("genero")["nota_publico"].agg(["count", "median", "min", "max"]))   # resumo por gênero
```

**Como ler:** **Drama** tem a mediana mais alta (8,55) e nenhuma nota abaixo de 8,4: é o gênero mais bem avaliado **e** mais consistente. **Comédia** tem a mediana mais baixa (7,40). **Ficção Científica** é o mais espalhado: vai de 7,0 (*Bacurau*) a 8,8 (*A Origem*).

**Cuidado:** a tabela mostra que cada gênero tem **poucos filmes** (de 2 a 6). Com grupos tão pequenos, um único filme muda muito o boxplot, então essas diferenças servem para praticar a leitura do gráfico, **não** para concluir que "drama é melhor que comédia".
