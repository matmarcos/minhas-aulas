# Parte 1 — Dois dados num jogo de tabuleiro

## 1. O espaço amostral

### A situação
Em vários jogos de tabuleiro (como o Banco Imobiliário), cada jogada é feita com **dois dados de 6 faces**, e tirar **faces iguais** (uma "dupla") dá direito a jogar de novo.

### A pergunta
Quais são **todos os resultados possíveis** de um lance de dois dados?

### A ferramenta da aula
O **espaço amostral** $\Omega$ é o conjunto de **todos os resultados possíveis** do experimento. Aqui, cada resultado é um **par** (dado 1, dado 2), com cada dado de 1 a 6. Pelo princípio multiplicativo, são $6 \times 6 = 36$ pares.

### O que o código faz
Monta a lista de todos os pares (d1, d2), fazendo d1 e d2 percorrerem de 1 a 6, e mostra quantos são.

```python
# todos os pares (d1, d2) com d1 de 1 a 6 e d2 de 1 a 6
espaco_amostral = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]

print(f"O espaço amostral tem {len(espaco_amostral)} elementos:")
print(espaco_amostral)
```

**Como ler:** são **36 pares**. Repare que (1, 2) e (2, 1) são resultados **diferentes**: no primeiro, o dado 1 deu 1; no segundo, deu 2. Como os dados são equilibrados, os 36 pares são **igualmente prováveis**.

## 2. Eventos e suas probabilidades

### A pergunta
Qual a chance de tirar uma **dupla** (faces iguais)? E de a soma dar **10**? E de a soma dar **menos que 2**?

### A ferramenta da aula
Um **evento** é um **subconjunto** do espaço amostral. Quando todos os resultados são igualmente prováveis:

$$P(A) = \frac{\text{nº de resultados em } A}{\text{nº de resultados em } \Omega}$$

### A resposta pela conta
- $A$ = "dupla" = {(1,1), (2,2), ..., (6,6)}: 6 pares, então $P(A) = 6/36 \approx 0{,}167$;
- $B$ = "soma 10" = {(4,6), (5,5), (6,4)}: 3 pares, então $P(B) = 3/36 \approx 0{,}083$;
- $C$ = "soma menor que 2": a menor soma possível é 1 + 1 = 2, então $C$ é o **evento vazio** e $P(C) = 0$.

### O que o código faz
Para cada evento, **filtra** do espaço amostral os pares que satisfazem a condição, mostra esses pares e divide a quantidade por 36.

```python
evento_A = [(d1, d2) for (d1, d2) in espaco_amostral if d1 == d2]          # faces iguais
evento_B = [(d1, d2) for (d1, d2) in espaco_amostral if d1 + d2 == 10]     # soma 10
evento_C = [(d1, d2) for (d1, d2) in espaco_amostral if d1 + d2 < 2]       # soma menor que 2

for nome, evento in [("A (dupla)", evento_A), ("B (soma 10)", evento_B), ("C (soma < 2)", evento_C)]:
    print(f"{nome}: {evento}")
    print(f"   P = {len(evento)}/36 = {len(evento) / 36:.3f}\n")
```

**Como ler:** a lista de cada evento mostra exatamente os pares que "contam". A chance de tirar uma dupla é **1 em 6** (0,167). O evento C não tem nenhum par: é o **evento impossível**, com probabilidade 0.

## 3. A conta bate com a prática?

### A pergunta
Se jogarmos os dados **de verdade** muitas vezes, a **frequência** de duplas fica perto de 1/6?

### A ferramenta da aula
A **frequência relativa** de um evento (quantas vezes ele aconteceu ÷ número de repetições) se aproxima da sua probabilidade quando o número de repetições é grande. É a interpretação **frequentista** da probabilidade.

### O que o código faz
Joga os dois dados **100.000 vezes** (`randint(1, 6)` sorteia um número inteiro de 1 a 6) e conta quantas vezes saiu dupla. No fim, divide pelo número de lançamentos.

```python
lancamentos = 100_000
duplas = 0

for _ in range(lancamentos):          # repete 100.000 vezes:
    d1 = random.randint(1, 6)         #   joga o dado 1
    d2 = random.randint(1, 6)         #   joga o dado 2
    if d1 == d2:                      #   se saiu dupla,
        duplas += 1                   #   conta mais uma

print(f"Frequência relativa de duplas (empírico): {duplas / lancamentos:.3f}")
print(f"Probabilidade pela contagem (teórico):    {6 / 36:.3f}")
```

**Como ler:** em 100.000 lances, a dupla saiu em **0,167** das vezes, a mesma probabilidade obtida contando os 36 pares. Contar os casos no espaço amostral **prevê** o que acontece na prática. Para o jogador: em média, uma a cada 6 jogadas dá direito a jogar de novo.

### A ferramenta da aula
O experimento é "sortear um filme". O **espaço amostral** é o conjunto dos **24 filmes**, e cada um tem probabilidade 1/24.

### O que o código faz
Digita o banco de dados dos filmes numa tabela. Cada lista entre colchetes é uma coluna, na mesma ordem dos títulos.

## 5. Qual a chance de sair um drama?

### A pergunta
Qual a probabilidade de o filme sorteado ser um **Drama**?

### A ferramenta da aula
O evento $D$ = "o filme é Drama" é o **subconjunto** dos filmes de drama. Como todos os filmes têm a mesma chance, $P(D) = \dfrac{\text{nº de dramas}}{24}$.

### O que o código faz
Filtra da tabela só as linhas com gênero "Drama" (`df[condição]` = "as linhas de df que satisfazem a condição"), conta quantas são e divide por 24.

```python
evento_drama = df[df["genero"] == "Drama"]          # só os filmes de drama

print(f"Espaço amostral: {len(df)} filmes")
print(f"Evento 'é Drama': {len(evento_drama)} filmes")
print(f"P(Drama) = {len(evento_drama)}/{len(df)} = {len(evento_drama) / len(df):.3f}\n")
print(evento_drama[["titulo", "genero"]].to_string(index=False))
```

**Como ler:** há **6 dramas** entre os 24 filmes, então $P(D) = 6/24 = 0{,}25$. A lista mostra quais são.

## 6. Combinando eventos: "e", "ou" e "não"

### A situação e a pergunta
No grupo há um adolescente de 15 anos, então os amigos querem saber a chance de sortear um filme **impróprio para ele** (classificação 16 ou 18 anos). Seja $I$ esse evento.
1. Qual a chance de sair um filme que é **drama e impróprio**?
2. Qual a chance de sair um filme que é **drama ou impróprio**?
3. Qual a chance de sair um filme que **não** é impróprio (todos podem assistir)?

### A ferramenta da aula
- **Interseção** $D \cap I$ ("e"): filmes que estão nos **dois** eventos;
- **União** $D \cup I$ ("ou"): filmes que estão em **pelo menos um** deles. Pela regra da adição, $P(D \cup I) = P(D) + P(I) - P(D \cap I)$: subtraímos a interseção para não contar duas vezes;
- **Complementar** $I^c$ ("não"): os filmes **fora** de $I$, com $P(I^c) = 1 - P(I)$.

### A resposta pela conta
São 8 filmes impróprios, e 4 deles são dramas. Então:
- $P(D \cap I) = 4/24 \approx 0{,}167$;
- $P(D \cup I) = 6/24 + 8/24 - 4/24 = 10/24 \approx 0{,}417$;
- $P(I^c) = 1 - 8/24 = 16/24 \approx 0{,}667$.

### O que o código faz
Cria duas colunas de **verdadeiro/falso** (é drama? é impróprio?) e as combina: `&` significa "e", `|` significa "ou" e `~` significa "não". Depois conta os filmes em cada caso.

```python
drama = df["genero"] == "Drama"                                   # True se o filme é drama
improprio = df["classificacao_indicativa"].isin(["16", "18"])     # True se é 16 ou 18 anos

print(f"P(Drama e Impróprio) = {(drama & improprio).sum()}/24 = {(drama & improprio).mean():.3f}")
print(f"P(Drama ou Impróprio) = {(drama | improprio).sum()}/24 = {(drama | improprio).mean():.3f}")
print(f"P(não Impróprio)      = {(~improprio).sum()}/24 = {(~improprio).mean():.3f}\n")
print("Dramas impróprios:", ", ".join(df[drama & improprio]["titulo"]))
```

**Como ler:** os números batem com a regra da adição e a do complementar. Se somássemos simplesmente $6 + 8 = 14$ filmes para o "ou", os 4 dramas impróprios (*O Poderoso Chefão*, *Coringa*, *Cidade de Deus* e *Parasita*) seriam contados **duas vezes**; por isso a regra subtrai a interseção. E há **2 chances em 3** de sair um filme que todo o grupo pode ver.

## 7. Conferindo por simulação

### A pergunta
Se o grupo sortear filmes muitas e muitas vezes, a **frequência** de dramas fica perto de 0,25?

### O que o código faz
Sorteia **100.000 filmes** do catálogo, **com reposição** (o filme sorteado volta para a lista; `replace=True`), e calcula a fração de sorteios que deram drama. O `random_state=42` é a semente desse sorteio.

```python
sorteios = df.sample(n=100_000, replace=True, random_state=42)     # 100.000 sorteios com reposição
freq_drama = (sorteios["genero"] == "Drama").mean()                # fração que deu drama

print(f"Frequência relativa de Drama (empírico): {freq_drama:.3f}")
print(f"P(Drama) pela contagem (teórico):        {6 / 24:.3f}")
```

**Como ler:** a frequência simulada (**0,251**) fica colada na probabilidade calculada (**0,250**), como aconteceu com os dados na Parte 1.
