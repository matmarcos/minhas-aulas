## 1. Reformulando o cenário como variável contínua

Suponha que o novo detector mede um nível de estresse X, uniformemente distribuído entre 0 e 10 (ou seja, qualquer valor nesse intervalo é igualmente provável). O funcionário é sinalizado como suspeito se X > 7.

```python
import numpy as np
import matplotlib.pyplot as plt

limite_inferior, limite_superior = 0, 10
limiar = 7

x = np.linspace(limite_inferior, limite_superior, 200)
densidade = np.full_like(x, 1 / (limite_superior - limite_inferior))  # densidade constante = 1/10

plt.plot(x, densidade, color="steelblue")
plt.fill_between(x[x > limiar], densidade[x > limiar], color="coral", alpha=0.6, label="X > 7 (sinalizado)")
plt.xlabel("Nível de estresse (X)")
plt.ylabel("Densidade")
plt.title("Densidade uniforme de X, com a área sinalizada em destaque")
plt.legend()
plt.show()
```

## 2. Calculando P(sinalizado) = P(X > 7)

Numa distribuição uniforme, a probabilidade de um intervalo é só a **área do retângulo** naquele intervalo — não precisa de integral complicada aqui, porque a densidade é constante.

```python
largura_intervalo_sinalizado = limite_superior - limiar
altura_densidade = 1 / (limite_superior - limite_inferior)
p_sinalizado = largura_intervalo_sinalizado * altura_densidade

print(f"Largura do intervalo (10 - 7): {largura_intervalo_sinalizado}")
print(f"Altura da densidade (1/10):    {altura_densidade}")
print(f"P(X > 7) = {largura_intervalo_sinalizado} × {altura_densidade} = {p_sinalizado:.2f}")
```

### Conferindo por simulação

```python
rng = np.random.default_rng(3)
n_funcionarios = 500_000

niveis_de_estresse = rng.uniform(limite_inferior, limite_superior, size=n_funcionarios)
fracao_sinalizada = (niveis_de_estresse > limiar).mean()

print(f"Fração sinalizada, simulada: {fracao_sinalizada:.3f}")
print(f"Valor teórico: {p_sinalizado:.3f}")
```

## 3. Valor esperado e variância da variável contínua

Para a distribuição uniforme entre a e b, existe fórmula pronta: E[X] = (a+b)/2 e Var[X] = (b-a)²/12. Vamos conferir se bate com a simulação.

```python
esperanca_teorica = (limite_inferior + limite_superior) / 2
variancia_teorica = (limite_superior - limite_inferior) ** 2 / 12

esperanca_simulada = niveis_de_estresse.mean()
variancia_simulada = niveis_de_estresse.var()

print(f"E[X] teórico:    {esperanca_teorica:.3f}   |   simulado: {esperanca_simulada:.3f}")
print(f"Var[X] teórico:  {variancia_teorica:.3f}   |   simulado: {variancia_simulada:.3f}")
```

Faz sentido o valor esperado dar exatamente no meio do intervalo (5) — numa distribuição uniforme, não há nenhum valor "mais provável" que outro, então o centro de gravidade da distribuição é o próprio ponto médio.
