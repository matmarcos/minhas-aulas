## 1. Probabilidade Total: P(falhou no teste)

Os mesmos números da Aula 5: 10% dos funcionários são ladrões; o teste acerta com probabilidade 0,8 em ambos os casos.

```python
p_ladrao = 0.10
p_honesto = 1 - p_ladrao

p_falhar_dado_ladrao = 0.8    # o teste flagra corretamente o ladrão com prob. 0,8
p_falhar_dado_honesto = 0.2   # o teste erra e flagra o honesto com prob. 0,2

# Teorema da Probabilidade Total: P(falhou) = P(falhou|ladrão)*P(ladrão) + P(falhou|honesto)*P(honesto)
p_falhou = p_falhar_dado_ladrao * p_ladrao + p_falhar_dado_honesto * p_honesto

print(f"P(falhou) = {p_falhar_dado_ladrao} × {p_ladrao} + {p_falhar_dado_honesto} × {p_honesto} = {p_falhou:.3f}")
```

Compare com a Aula 5: simulamos e encontramos cerca de 26% dos funcionários falhando no teste — bate exatamente com o valor teórico calculado aqui.

## 2. Teorema de Bayes: P(ladrão | falhou)

Agora a pergunta que os "gerentes" fizeram: de quem falhou no teste, qual fração é realmente ladrão?

```python
# Teorema de Bayes: P(ladrão | falhou) = P(falhou | ladrão) * P(ladrão) / P(falhou)
p_ladrao_dado_falhou = (p_falhar_dado_ladrao * p_ladrao) / p_falhou

print(f"P(ladrão | falhou) = ({p_falhar_dado_ladrao} × {p_ladrao}) / {p_falhou:.3f} = {p_ladrao_dado_falhou:.3f}")
print(f"\nOu seja: {p_ladrao_dado_falhou:.1%} — bate com o ~31% que a simulação da Aula 5 encontrou.")
```

## 3. O exemplo das duas urnas, da aula

Uma urna tem 3 brancas e 2 amarelas; a outra tem 4 brancas e 2 amarelas. Escolhe-se uma urna ao acaso (50%/50%) e dela retira-se uma bola. Qual a probabilidade de a bola ser branca?

```python
p_urna1 = 0.5
p_urna2 = 0.5
p_branca_dado_urna1 = 3 / 5
p_branca_dado_urna2 = 4 / 6

p_branca = p_branca_dado_urna1 * p_urna1 + p_branca_dado_urna2 * p_urna2
print(f"P(branca) = {p_branca_dado_urna1:.3f} × {p_urna1} + {p_branca_dado_urna2:.3f} × {p_urna2} = {p_branca:.3f}")
```

```python
import numpy as np
rng = np.random.default_rng(1)
n = 300_000

urna_escolhida = rng.integers(1, 3, size=n)  # 1 ou 2, cada uma com 50% de chance
eh_branca = np.where(
    urna_escolhida == 1,
    rng.random(n) < p_branca_dado_urna1,
    rng.random(n) < p_branca_dado_urna2,
)
print(f"P(branca), simulado: {eh_branca.mean():.3f}")
```

## 4. Bônus de curiosidade: o problema de Monty Hall

O vídeo recomendado desta aula usa a simulação do problema de Monty Hall como aplicação do Teorema de Bayes. O problema: você escolhe uma de 3 portas (atrás de uma tem um prêmio, as outras duas têm nada). O apresentador, que sabe onde está o prêmio, abre uma das portas que você não escolheu e que não tem o prêmio, e te pergunta se você quer trocar de porta. **Vale a pena trocar?**

Isso não é obrigatório para a disciplina — é só uma curiosidade clássica de probabilidade.

```python
n_jogos = 100_000
portas = np.array([0, 1, 2])

premio = rng.integers(0, 3, size=n_jogos)          # porta com o prêmio, sorteada em cada jogo
escolha_inicial = rng.integers(0, 3, size=n_jogos)  # porta que o jogador escolhe primeiro

# se ficar com a escolha inicial, ganha só quando a escolha inicial já era a porta certa
ganha_ficando = (escolha_inicial == premio)

# se trocar, ganha exatamente quando a escolha inicial NÃO era a porta certa
# (porque o apresentador sempre abre uma porta errada das duas restantes,
#  sobrando só a porta certa para trocar, quando a escolha inicial estava errada)
ganha_trocando = ~ganha_ficando

print(f"Taxa de vitória ficando com a escolha inicial: {ganha_ficando.mean():.1%}")
print(f"Taxa de vitória trocando de porta:              {ganha_trocando.mean():.1%}")
```

**Trocar de porta dobra sua chance de ganhar** (de 1/3 para 2/3) — é o mesmo tipo de raciocínio contraintuitivo do detector de mentiras: a informação nova (o apresentador abrir uma porta) muda a probabilidade condicional, mesmo que pareça que "não deveria fazer diferença".
